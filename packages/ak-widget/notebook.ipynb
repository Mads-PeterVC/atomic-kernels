```python
from ak_widget import ViewerWidget
from ase.build import molecule, bulk
from ase.collections import g2
```

```python
trajectory = []
for name in g2.names:    
    _atoms = molecule(name)
    _atoms.set_cell([10, 10, 10])
    _atoms.center()
    trajectory.append(_atoms)

viewer = ViewerWidget(trajectory)
viewer
```

```python
bulk_trajectory = []
for element in ['Al', 'Cu', 'Fe', 'Si']:
    _atoms = bulk(element, 'fcc', a=4.05)
    bulk_trajectory.append(_atoms)


viewer_1 = ViewerWidget(bulk_trajectory)
viewer_1
```

```python
from ase.build import (
    fcc100, fcc110, fcc111, fcc211, fcc111_root,
    bcc100, bcc110, bcc111, bcc111_root,
    hcp0001, hcp10m10, hcp0001_root,
    diamond100, diamond111
)

surface_trajectory = []
for surface in [fcc100, fcc110, fcc111, fcc211,
                bcc100, bcc110, bcc111,
                hcp0001, hcp10m10,
                diamond100, diamond111]:
    try:
        _atoms = surface('Cu', size=(2, 2, 3), vacuum=10)
        surface_trajectory.append(_atoms)
    except Exception as e:
        pass

viewer_2 = ViewerWidget(surface_trajectory)
viewer_2
```
